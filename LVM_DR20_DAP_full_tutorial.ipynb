
# LVM DR20 DAP Tutorial (Publication-level)

This notebook provides a comprehensive, publication-level tutorial on how to use the **LVM Data Analysis Pipeline (DAP) DR20 products**, using the **Helix Nebula** as a working example while keeping the workflow fully general.

It follows the pedagogical structure of the DR19 tutorial, updated for DR20 improvements.


```

import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
```


## 1. Load a DAP file

Replace the filename with your DR20 DAP file.


```

filename = "dap_example.fits"
hdul = fits.open(filename)
hdul.info()
```


## 2. DAP Workflow (DR20)

The DR20 DAP follows a multi-step procedure:

1. Non-linear stellar fitting (velocity, dispersion, extinction)
2. Initial emission-line fitting
3. Full stellar population decomposition
4. Non-parametric emission-line measurements
5. Refined emission-line fitting

This allows separation of stellar and gas contributions and robust extraction of physical parameters.



## 3. Structure of the DAP file

Main extensions:

- PT: fiber positions
- RSP: stellar parameters
- COEFFS: stellar population decomposition
- PM_ELINES: parametric emission lines
- NP_ELINES_*: non-parametric emission lines
- ELINES_SIGMA_CHI: velocity dispersion and χ²
- ELINES_CHI2_AVG: average χ²
- INFO: configuration parameters


```

pt = hdul['PT'].data
rsp = hdul['RSP'].data
elines = hdul['NP_ELINES_R'].data
```


## 4. Example: Stellar mass distribution


```

if 'log_Mass' in rsp.names:
    plt.hist(rsp['log_Mass'], bins=30)
    plt.xlabel('log(Mass)')
    plt.ylabel('N')
    plt.show()
```


## 5. Emission-line analysis

Two approaches:

- Parametric: Gaussian fits for strong lines
- Non-parametric: flux moments for many lines

Use NP_ELINES_* for general analysis.



## 6. Spatial maps

Reconstruct maps using fiber positions.


```

if 'Ha_6562.85' in elines.names:
    x = pt['ra']
    y = pt['dec']
    z = elines['Ha_6562.85']
    plt.scatter(x, y, c=z, s=5)
    plt.colorbar(label='Hα flux')
    plt.xlabel('RA')
    plt.ylabel('DEC')
    plt.show()
```


## 7. Quality diagnostics (DR20)

- ELINES_SIGMA_CHI: velocity dispersion and χ² per line
- ELINES_CHI2_AVG: global fitting quality

These allow filtering unreliable measurements.



## 8. Configuration (INFO extension)

The INFO extension records the full configuration used by the DAP, enabling reproducibility.


```

info = hdul['INFO'].data
print(info)
```
